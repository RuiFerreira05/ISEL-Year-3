```python
import numpy as np
import numpy.random as rd
import matplotlib.pyplot as plt
import sklearn.datasets as dt
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, precision_recall_curve
import pickle
from sklearn.ensemble import RandomForestClassifier
import scipy.spatial.distance as spd
import sklearn.preprocessing as pp
from sklearn.decomposition import PCA
import numpy.linalg as la
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.feature_extraction.text import CountVectorizer
```

```python
D = dt.load_breast_cancer()

X = D.data
y = 1 - D.target
```

```python
X1, X2, y1, y2 = train_test_split(X, y, test_size=1/3, random_state=42, stratify=y)
```

```python
svm = SVC().fit(X1, y1)
```

```python
y2e = svm.predict(X2)
cm = confusion_matrix(y2, y2e)
cm
```

```python
sc = StandardScaler().fit(X1)
X1s = sc.transform(X1)
sc = StandardScaler().fit(X2)
X2s = sc.transform(X2)
```

```python
svm_std = SVC().fit(X1s, y1)
y2e_std = svm_std.predict(X2s)
cm_std = confusion_matrix(y2, y2e_std)
cm_std
```
