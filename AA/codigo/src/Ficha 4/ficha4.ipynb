```python
with open('./A51597_Q001_data.p', 'rb') as fs:
    data = pickle.load(fs)
    
print(data.keys())

X = data['x']
y = data['y']
folds = data['folds']
```

```python
X_train = X[folds == 1]
y_train = y[folds == 1]
X_test = X[folds == 0]
y_test = y[folds == 0]
```

```python
# regressão linear de 3º grau
poly = PolynomialFeatures(degree=3)
X_train_poly = poly.fit_transform(X_train.reshape(-1, 1))
X_test_poly = poly.transform(X_test.reshape(-1, 1))
lr = LinearRegression()
lr.fit(X_train_poly, y_train)
y_pred = lr.predict(X_test_poly)
mse = np.mean((y_test - y_pred) ** 2)
print(f'Mean Squared Error: {mse:.2f}')

# R2

r2 = lr.score(X_test_poly, y_test)
print(f'R^2 Score: {r2:.2f}')
```

```python
with open('./A51597_Q003_data.p', 'rb') as f:
    data = pickle.load(f)

print(data.keys())

X = data['X']
y = data['y']

print(f'X shape: {X.shape}\ny shape: {y.shape}')
```

```python
# y_pred = w.T @ X = w0 + w1*x1 + w2*x2 X belongs to w1_pred for y_pred >= 0
# and w = [w0, w1, w2].T and X = [1, x1, x2].T

# w_mse is the weights that minimize the mean squared error
# 1/N * sum((y - y_pred)^2)

# n = 1, ..., N.
# y[n] = -1 for X[n] belonging to w0 and y[n] = +1 for X[n] belonging to w1

# lr = LogisticRegression()
# lr.fit(X.T, y)
# lr.intercept_ = np.array([0.00])
# lr.coef_ = np.array([0.90, -0.43])

w = np.array([0.00, 0.90, -0.43])
X_bias = np.vstack((np.ones(X.shape[1]), X))
Rx = w @ X_bias
y_pred = (Rx >= 0)

cm = confusion_matrix(y, y_pred)

print(cm)

# precisao

precision = np.sum(y_pred[y == 1] == 1) / np.sum(y_pred == 1)
print(f'Precision: {precision:.2f}')
```

```python
lr = LogisticRegression()
lr.fit(X.T, y)
lr.intercept_ = np.array([0.00])
lr.coef_ = np.array([0.87, -0.50])

y_pred = lr.predict(X.T)

cm = confusion_matrix(y, y_pred)
print(cm)

# recall
recall = np.sum(y_pred[y == 1] == 1) / np.sum(y == 1)
print(f'Recall: {recall:.3f}')

# total acertos
total_right = np.sum(y_pred == y)
print(f'Total correct predictions: {total_right}')

# false positives rate
fpr = np.sum(y_pred[y == 0] == 1) / np.sum(y == 0)
print(f'False Positive Rate: {fpr:.3f}')

# n de erros em w0
n_errors_w0 = np.sum(y_pred[y == 0] == 1)
print(f'Number of errors in w0: {n_errors_w0}')
```
